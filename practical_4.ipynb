```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
# Simulate a coin toss experiment
num_tosses = 10000  # Number of coin tosses
results = np.random.choice(['Heads', 'Tails'], size=num_tosses)

# Count occurrences
num_heads = np.sum(results == 'Heads')
num_tails = np.sum(results == 'Tails')

# Compute probabilities
probability_heads = num_heads / num_tosses
probability_tails = num_tails / num_tosses

print(f"Total tosses: {num_tosses}")
print(f"Number of Heads: {num_heads}")
print(f"Number of Tails: {num_tails}")
print(f"Probability of Heads: {probability_heads:.4f}")
print(f"Probability of Tails: {probability_tails:.4f}")
```

```python
# Generate data for a normal distribution
mu, sigma = 0, 0.1 # mean and standard deviation
sample_size = 1000
s = np.random.normal(mu, sigma, sample_size)

# Plot the histogram
plt.hist(s, bins=50, density=True, alpha=0.6, color='g')

# Plot the PDF for comparison
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
plt.plot(x, stats.norm.pdf(x, mu, sigma), linewidth=2, color='r')

plt.title('Histogram of a Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency / Density')
plt.grid(True)
plt.show()
```
